This is a test Jupyter Notebook for connecting to kernel.
Fun fact: A kernel is the overall term for the computational engine that reads, compiles, and executes the code inside an interactive, persistent workspace that keeps your variables alive in memory while you experiment. Ex: Jupyter notebook, interactive shell.
TLDR: kernels are what allow you to run Python inside a Jupyter Notebook

Traditional software and programming does not use a kernel. 

Typically a interpreter will read the code and execute all of it right away, closing and erasing all memory after.

A kernel allows the interpreter, compiler, and executer to essentially stay open for a jupyter notebook without deleting any variables or memory we might use in later code cells.
