`````python
import subprocess
import sys

# List of required libraries
required_libraries = ["pandas", "google-cloud-bigquery", "google-cloud-storage", "google-auth", "google-cloud-bigquery"]

# Function to check and install missing libraries
def check_and_install_library(library):
    try:
        __import__(library.split("-")[0])  # Check if the library is installed
        print(f"✅ Library '{library}' is already installed.")
    except ImportError:
        print(f"❌ Library '{library}' is missing.")
        print(f"Installing '{library}'...")
        try:
            subprocess.check_call([sys.executable, "-m", "pip", "install", library])
            print(f"🎉 Successfully installed '{library}'!")
        except Exception as e:
            print(f"🚨 Error installing '{library}': {e}")
            print("👉 Try running the following command in your terminal:")
            print(f"pip install {library}")

# Check and install libraries
for lib in required_libraries:
    check_and_install_library(lib)
`````

# Step 3: Uploading Data from SQLite to BigQuery

## Why Upload Data to BigQuery?
BigQuery is a powerful data warehouse that allows us to:
1. Store large datasets in the cloud.
2. Query data using SQL with incredible speed and scalability.
3. Integrate with tools like Looker Studio for visualization.

In this step, we’ll:
1. Extract data from an SQLite database.
2. Transform it into a format BigQuery understands(Pandas DataFrame).
3. Upload it directly to a BigQuery table.

---

## 🔗  References
- [BigQuery Documentation](https://cloud.google.com/bigquery/docs)
- [Google BigQuery Python Client Library](https://cloud.google.com/python/docs/reference/bigquery/latest)
- [SQLite Documentation](https://www.sqlite.org/docs.html)

`````python
#Import Required Libraries and Set Up Configuration
import pandas as pd
from sqlite3 import connect
from google.cloud import bigquery
from google.oauth2 import service_account

# Set up the path to your SQLite database and BigQuery credentials
SQLITE_DB_FILE = r"C:\Users\stazm\OneDrive\Downloads\DAEN_328\ASGMT_1\nyc_covid_outcomes.db"  # Replace with your SQLite database file from the first assignment
BIGQUERY_KEY_FILE = r"C:\Users\stazm\OneDrive\Downloads\DAEN_328\bigquery-key.json"  # Replace with your Service Account JSON key file that you just downloaded
PROJECT_ID = "daen328assignmentbigquery"  # Replace with your GCP Project ID
DATASET_ID = "nyc_covid_outcomes"  # Replace with your BigQuery dataset ID(this is the name of the dataset in the previous step)
# PROJECT_ID: 
# - Found in the GCP Console (https://console.cloud.google.com)
# - Navigate to the top-left corner, select the project (id must be displayed next to the project name or on the homescreen)
# DATASET_ID:
# - Found under "BigQuery" in the GCP Console.
# - Navigate to the BigQuery section (https://console.cloud.google.com/bigquery).
# - Select your project from the left panel, then select your dataset to view the dataset info 
`````

###  Loading Data from SQLite:
1. Connect to the SQLite database.
2. Use a SQL query to extract all records from the `covid_outcomes` table.
3. Load the extracted data into a Pandas DataFrame for further processing.

`````python
# Load Data from SQLite into a Pandas DataFrame
def load_data_from_sqlite():
    """
    Load data from the 'covid_outcomes' table into a Pandas DataFrame.
    
    Returns:
        pd.DataFrame: Data loaded from the SQLite table.
    """
    try:
        # Connect to the SQLite database
        conn = connect(SQLITE_DB_FILE)
        query = "SELECT * FROM covid_outcomes"
        print("🔄 Loading data from the 'covid_outcomes' table...")
        data = pd.read_sql_query(query, conn)
        conn.close()
        print(f"✅ Successfully loaded {len(data)} records from 'covid_outcomes'.")
        return data
    except Exception as e:
        print(f"❌ Error loading data from 'covid_outcomes': {e}")
        return pd.DataFrame()
`````

### Uploading Data to BigQuery
1. Authenticate with BigQuery using the Service Account key.
2. Define a schema for the BigQuery table (if it doesn’t already exist).
3. Upload the data from the DataFrame into the BigQuery table.


`````python
#Upload DataFrame to BigQuery
def upload_to_bigquery(dataframe, table_name):
    """
    Upload a Pandas DataFrame to a BigQuery table.
    
    Args:
        dataframe (pd.DataFrame): The DataFrame to upload.
        table_name (str): The name of the BigQuery table (e.g., "your_dataset_id.table_name").
    """
    try:
        # Authenticate and initialize the BigQuery client
        credentials = service_account.Credentials.from_service_account_file(BIGQUERY_KEY_FILE)
        client = bigquery.Client(credentials=credentials, project=PROJECT_ID)

        # Define table ID
        table_id = f"{PROJECT_ID}.{DATASET_ID}.{table_name}"
        
        # Configure the load job
        job_config = bigquery.LoadJobConfig(
            write_disposition=bigquery.WriteDisposition.WRITE_APPEND,  # Append to existing table
            create_disposition=bigquery.CreateDisposition.CREATE_IF_NEEDED  # Create table if not exists
        )

        # Start the upload job
        print(f"🔄 Uploading {len(dataframe)} records to BigQuery table: {table_id}")
        job = client.load_table_from_dataframe(dataframe, table_id, job_config=job_config)
        job.result()  # Wait for the job to complete
        print(f"✅ Successfully uploaded {len(dataframe)} records to {table_id}")
    except Exception as e:
        print(f"❌ Error uploading to BigQuery: {e}")
`````

`````python
# Full Workflow
def push_to_bigquery():
    """
    Full workflow to load data from SQLite and upload to BigQuery.
    """
    print("🔄 Starting the workflow to upload data to BigQuery...")
    # Load data from SQLite
    dataframe = load_data_from_sqlite()
    
    if not dataframe.empty:
        # Specify the table name
        table_name = "covid_outcomes"  # Replace with your desired BigQuery table name
        
        # Upload data to BigQuery
        upload_to_bigquery(dataframe, table_name)
    else:
        print("⚠️ No data found in the 'covid_outcomes' table. Workflow skipped.")

# Execute the workflow
push_to_bigquery()
`````
