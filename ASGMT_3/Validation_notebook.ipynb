## ✏️ Normalizing Text Data

Text fields often contain **special characters or inconsistent casing**. The `normalize_text` function standardizes text for uniformity.

### 🛠️ How It Works:

- **Removes special characters** (e.g., punctuation, symbols).
- **Converts text to lowercase** for consistency.

### ✅ Example:

```
Input:  "Hello, World!"
Output: "hello world"
```




````python
# --------------------------------------------------
# Cell 4.1: normalize_text
# --------------------------------------------------
def normalize_text(text):
    """
    Removes special characters and converts text to lowercase.
    """
    import re
    return re.sub(r"[^a-zA-Z0-9\s]", "", str(text)).lower()
````

## 🆔 Validating Employee ID

Employee IDs must be **unique, non-null, and non-empty** to ensure proper record tracking.

### 🛠️ How It Works:

- Checks if `EmployeeID` is missing or empty.
- Flags duplicates as errors.

### ⚠️ Potential Issue:

EmployeeID must be unique and cannot be null or empty.

````python
# --------------------------------------------------
# Cell 4.2: validate_employee_id
# --------------------------------------------------
def validate_employee_id(row, errors, is_duplicate):
    """
    Checks EmployeeID is non-null, non-empty, and not duplicated.
    """
    import pandas as pd
    if pd.isna(row['EmployeeID']) or row['EmployeeID'] == "" or is_duplicate:
        errors.append("EmployeeID must be unique and cannot be null or empty.")
````

## 📝 Validating Name

Names should be **properly formatted, non-null, and unique** to maintain data consistency.

### 🛠️ How It Works:

- Ensures `Name` is not missing or empty.
- Flags duplicate names as errors.
- Checks if the name follows **title case** (e.g., `John Doe`).

### ⚠️ Potential Issues:

```
Name cannot be null, empty, or duplicated.
Name must be in title case (e.g., 'John Doe').
```



````python
# --------------------------------------------------
# Cell 4.3: validate_name
# --------------------------------------------------
def validate_name(row, errors, is_duplicate):
    """
    Checks Name is non-null, not duplicated, and in title case.
    """
    import pandas as pd
    if pd.isna(row['Name']) or row['Name'] == "" or is_duplicate:
        errors.append("Name cannot be null, empty, or duplicated.")
    elif not row['Name'].istitle():
        errors.append("Name must be in title case (e.g., 'John Doe').")
````

## 🎓 Validating Highest Degree

````python
# --------------------------------------------------
# Cell 4.4: validate_highest_degree
# --------------------------------------------------
def validate_highest_degree(row, errors):
    """
    Checks HighestDegree is non-null and non-empty.
    """
    import pandas as pd
    if pd.isna(row['HighestDegree']) or row['HighestDegree'] == "":
        errors.append("HighestDegree cannot be null or empty.")
````

## 🏢 Validating Division

The `Division` field must be **valid and match its corresponding address** to ensure accurate department records.

### 🛠️ How It Works:

- Ensures `Division` is **not null or empty**.
- Confirms that `Division` exists in the predefined `division_names` list.
- Verifies that `NormalizedDivisionAddress` matches the expected value for the division.

### Potential Issues:

```
Division cannot be null or empty.
Invalid Division: [Division]. Must be one of [division_names].
Invalid DivisionAddress for [Division].
```



````python
# --------------------------------------------------
# Cell 4.5: validate_division
# --------------------------------------------------
def validate_division(row, errors, division_names, division_addresses_normalized):
    """
    Checks Division is valid and matches the normalized division address.
    """
    import pandas as pd
    if pd.isna(row['Division']) or row['Division'] == "":
        errors.append("Division cannot be null or empty.")
    elif row['Division'] not in division_names:
        errors.append(f"Invalid Division: {row['Division']}. Must be one of {division_names}.")
    else:
        if row['NormalizedDivisionAddress'] != division_addresses_normalized.get(row['Division'], ""):
            errors.append(f"Invalid DivisionAddress for {row['Division']}.")
````

## 🏢 Validating Department

The `Department` field must be **valid and part of the predefined department list** to ensure proper categorization.

### 🛠️ How It Works:

- Ensures `Department` is **not null or empty**.
- Confirms that `Department` exists in the predefined `valid_departments` list.

### Potential Issues:

```
Department cannot be null or empty.
Invalid Department: [Department].
```


````python
# --------------------------------------------------
# Cell 4.6: validate_department
# --------------------------------------------------
def validate_department(row, errors, valid_departments):
    """
    Checks Department is non-null and in the valid departments list.
    """
    import pandas as pd
    if pd.isna(row['Department']) or row['Department'] == "":
        errors.append("Department cannot be null or empty.")
    elif row['Department'] not in valid_departments:
        errors.append(f"Invalid Department: {row['Department']}.")
````

## 📧 Validating Email

The `Email` field must follow a **strict format** to ensure consistency and correctness.

### 🛠️ How It Works:

- Uses a **regular expression** to verify the format.
- Ensures emails follow the pattern: `firstname_lastname_number@KLB.com`.

### Potential Issue:

```
Email must be in format 'firstname_lastname_number@KLB.com'.
```


````python
# --------------------------------------------------
# Cell 4.8: validate_email
# --------------------------------------------------
def validate_email(row, errors):
    """
    Ensures Email matches the firstname_lastname_number@KLB.com format.
    """
    import re
    pattern = r"^[a-z]+_[a-z]+_\d+@KLB\.com$"
    if not re.match(pattern, str(row['Email'])):
        errors.append("Email must be in format 'firstname_lastname_number@KLB.com'.")
````

## 📞 Validating Phone Number

The `PhoneNumber` field must follow a **standardized format** to ensure uniformity.

### 🛠️ How It Works:

- Uses a **regular expression** to check the correct pattern.
- Ensures phone numbers follow the format `(XXX)XXX-XXXX`.

### Potential Issue:

```
Invalid PhoneNumber, PhoneNumber must be in format (XXX)XXX-XXXX.
```



````python
# --------------------------------------------------
# Cell 4.9: validate_phone_number
# --------------------------------------------------
def validate_phone_number(row, errors):
    """
    Ensures PhoneNumber is in the (XXX)XXX-XXXX format.
    """
    import re
    pattern = r"^\(\d{3}\)\d{3}-\d{4}$"
    if not re.match(pattern, str(row['PhoneNumber'])):
        errors.append("Invalid PhoneNumber, PhoneNumber must be in format (XXX)XXX-XXXX.")
````

## 🔐 Validating Social Security Number

The `SocialSecurity` field must be **unique, non-null, and properly formatted** to maintain data integrity.

### 🛠️ How It Works:

- Ensures `SocialSecurity` is **not empty or duplicated**.
- Uses a **regular expression** to check the format `XXX-XX-XXXX`.

### Potential Issues:

```
SocialSecurity must be unique and cannot be null or empty.
Invalid SocialSecurity format. Must be XXX-XX-XXXX.
```



````python
# --------------------------------------------------
# Cell 4.10: validate_social_security
# --------------------------------------------------
def validate_social_security(row, errors, is_duplicate):
    """
    Checks that SocialSecurity is unique and matches XXX-XX-XXXX.
    """
    import pandas as pd
    import re
    if pd.isna(row['SocialSecurity']) or row['SocialSecurity'] == "" or is_duplicate:
        errors.append("SocialSecurity must be unique and cannot be null or empty.")
    elif not re.match(r"^\d{3}-\d{2}-\d{4}$", str(row['SocialSecurity'])):
        errors.append("Invalid SocialSecurity format. Must be XXX-XX-XXXX.")
````

## 💰 Validating Salary

The `Salary` field must be **a valid number** and fall within the expected range for the given job title.

### 🛠️ How It Works:

- Ensures `JobTitle` is **not empty**.
- Checks if the job title exists in `salary_ranges_df`.
- Verifies if `Salary` falls within the **minimum and maximum** salary range for that job title.

### Potential Issues:

```
JobTitle did not match or is empty.
No salary range found for job title: [JobTitle].
Salary out of range for [JobTitle]. Should be between [MinSalary] and [MaxSalary].
Salary must be a number.
```



````python
# --------------------------------------------------
# Cell 4.11: validate_salary
# --------------------------------------------------
def normalize_job_title(job_title):
    """
    Standardizes job titles by replacing 'Human Resources' with 'HR' 
    and ensuring consistent capitalization.
    """
    if pd.isna(job_title) or job_title.strip() == "":
        return None
    return job_title.replace("Human Resources", "HR").replace("Hr", "HR").strip()

def validate_salary(row, errors, salary_ranges_df):
    """
    Checks job title validity and salary range, using normalized job titles without modifying salary_ranges_df.
    """
    import pandas as pd

    job_title = normalize_job_title(row['JobTitle'])

    if not job_title:
        errors.append("JobTitle did not match or is empty.")
        return

    # Retrieve salary range using a dynamically normalized comparison
    job_salary_range = salary_ranges_df[salary_ranges_df['JobTitle'].apply(normalize_job_title) == job_title]

    if not job_salary_range.empty:
        min_salary, max_salary = job_salary_range.iloc[0][['MinSalary', 'MaxSalary']]
        try:
            salary = float(row['Salary'])
            if not (min_salary <= salary <= max_salary):
                errors.append(f"Salary out of range for {row['JobTitle']}. Should be between {min_salary} and {max_salary}.")
        except ValueError:
            errors.append("Salary must be a number.")
    else:
        errors.append(f"No salary range found for job title: {row['JobTitle']}.")

````

## 📅 Validating Birthdate

The `Birthdate` field must follow the **MM/DD/YYYY** format with leading zeros when needed.

### 🛠️ How It Works:

- Uses a **regular expression** to check the correct format.
- Normalizes single-digit months and days to two-digit format (e.g., `1/5/2000` → `01/05/2000`).

### Potential Issue:

```
Birthdate must be in MM/DD/YYYY format.
```



````python
# --------------------------------------------------
# Cell 4.12: validate_birthdate
# --------------------------------------------------
def validate_birthdate(row, errors):
    """
    Ensures birthdate follows MM/DD/YYYY format with leading zeros if needed.
    """
    import re

    def normalize_birthdate(birthdate_str):
        match_obj = re.match(r"^(\d{1,2})/(\d{1,2})/(\d{4})$", birthdate_str.strip())
        if match_obj:
            month, day, year = match_obj.groups()
            return f"{int(month):02d}/{int(day):02d}/{year}"
        else:
            return None

    birthdate_str = str(row['Birthdate']).strip()
    normalized = normalize_birthdate(birthdate_str)
    if normalized:
        row['Birthdate'] = normalized
    else:
        errors.append("Birthdate must be in MM/DD/YYYY format.")
````

## ✅ Main Validation & Logging

This is the **core validation process** that checks each row against multiple data quality rules.

### 🛠️ How It Works:

1. **Prepares Data:**

   - Detects **duplicate values** for `EmployeeID`, `Name`, and `SocialSecurity`.
   - Normalizes **DivisionAddress** for verification.
   - Extracts **valid departments** from `job_desc_df`.

2. **Validates Each Row:**

   - Runs a series of validation functions:
     - `validate_employee_id`
     - `validate_name`
     - `validate_highest_degree`
     - `validate_division`
     - `validate_department`
     - `validate_email`
     - `validate_phone_number`
     - `validate_social_security`
     - `validate_salary`
     - `validate_birthdate`
   - Collects any errors found.

3. **Error Handling & Logging:**

   - Errors are stored along with the **row number** and **EmployeeID**.
   - If errors exist:
     - They are displayed with details.
     - An **error summary** is generated.
     - A log file is created with a timestamp.

### 📋 Understanding the Output:

#### **Validation Errors Found:**

- Displays all **validation errors** detected in the dataset.
- Each error is associated with a **specific row and EmployeeID**.
- Helps identify **what needs to be corrected** for data integrity.

```
🔴 Validation Errors Found:
Row 5 (EmployeeID: 12345):
   - Name must be in title case.
   - Salary out of range.
```

#### **Error Summary:**

- Aggregates the **frequency of each error** for quick analysis.
- Helps **prioritize fixes** based on the most common issues.

```
🔎 Error Summary:
Error Message                          Count
--------------------------------------  -----
Name must be in title case.            3
Salary out of range.                   2
```

**📏 Graded count:** the rubric's *\"fewer than 50 errors\"* means **Total Error Messages** — the sum of the `Count` column (one row can contribute several errors). It is printed at the end of the report and written at the bottom of the log file.

#### **Log File Generation:**

- Saves error details in a timestamped log file (`validation_log_YYYYMMDD_HHMMSS.txt`) in the same directory.
- Useful for **debugging issues**.

#### **If No Errors Are Found:**

- Confirms that the dataset meets validation criteria.
- Indicates that **no corrections** are needed.

```
✅ No validation errors found! Great job!
```

#### **If DataFrames Are Missing:**

- Warns that validation could not be completed due to missing datasets.
- Prevents running validation on incomplete or incorrect data.

```
⚠ Could not perform validation because one or more DataFrames are invalid.
```

This ensures **data accuracy, consistency, and completeness** before further processing.



````python
# --------------------------------------------------
# Cell 5: Main Validation & Logging
# --------------------------------------------------

def validate_student_csv(student_df, job_desc_df, salary_ranges_df):
    if student_df is None or student_df.empty:
        print("No student data to validate.")
        return None
    
    # Reset index if needed and prepare duplication checks
    student_df = student_df.reset_index(drop=True)
    dup_employee_series = student_df.duplicated(subset=['EmployeeID'])
    dup_name_series = student_df.duplicated(subset=['Name'])
    dup_ssn_series = student_df.duplicated(subset=['SocialSecurity'])
    
    # Create normalized addresses for division checks
    division_addresses = {
        'North Division': '123 North St, Springfield, IL 62701',
        'South Division': '456 South Ave, Austin, TX 73301',
        'East Division': '789 East Blvd, Albany, NY 12201',
        'West Division': '101 West Dr, Sacramento, CA 94203',
        'Central Division': '202 Main St, Columbus, OH 43085'
    }
    division_names = set(division_addresses.keys())
    
    student_df['NormalizedDivisionAddress'] = student_df['DivisionAddress'].apply(
        lambda x: normalize_text(x) if pd.notna(x) else ""
    )
    division_addresses_normalized = {k: normalize_text(v) for k, v in division_addresses.items()}

    valid_departments = set()
    if job_desc_df is not None and not job_desc_df.empty:
        valid_departments = set(job_desc_df['Department'].unique())

    error_log = []
    
    # Validate each row
    for idx, row in student_df.iterrows():
        errors = []
        
        # Call individual validation functions
        validate_employee_id(row, errors, dup_employee_series.loc[idx])
        validate_name(row, errors, dup_name_series.loc[idx])
        validate_highest_degree(row, errors)
        validate_division(row, errors, division_names, division_addresses_normalized)
        validate_department(row, errors, valid_departments)
        validate_email(row, errors)
        validate_phone_number(row, errors)

           ###Above is chill (employee ID later and easy)

        validate_social_security(row, errors, dup_ssn_series.loc[idx])
        validate_salary(row, errors, salary_ranges_df)
        validate_birthdate(row, errors)

        if errors:
            # Store row index (1-based) + the EmployeeID + the error messages
            error_log.append([idx + 1, row['EmployeeID'], errors])
    
    # Convert errors to DataFrame
    error_df = pd.DataFrame(error_log, columns=["Row", "EmployeeID", "Errors"])
    return error_df


# -----------------------------
# Actually perform validation:
# -----------------------------
if (
    student_df is not None and not student_df.empty 
    and job_desc_df is not None 
    and salary_ranges_df is not None
):
    error_df = validate_student_csv(student_df, job_desc_df, salary_ranges_df)
    
    if error_df is not None and not error_df.empty:
        print("\n🔴 Validation Errors Found:\n")
        for _, row in error_df.iterrows():
            print(f"Row {row['Row']} (EmployeeID: {row['EmployeeID']}):")
            for e in row['Errors']:
                print(f"   - {e}")
            print()

        # Summarize the errors
        error_df['Error_List'] = error_df['Errors']
        exploded = error_df.explode('Error_List')
        summary = exploded['Error_List'].value_counts().reset_index()
        summary.columns = ['Error Message', 'Count']
        
        print("🔎 Error Summary:\n")
        print(summary.to_string(index=False))
        total_error_messages = int(summary['Count'].sum())
        
        # Save error log
        log_filename = f"validation_log_{datetime.now().strftime('%Y%m%d_%H%M%S')}.txt"
        with open(log_filename, 'w') as f:
            for _, rowx in error_df.iterrows():
                f.write(f"Row {rowx['Row']} (EmployeeID: {rowx['EmployeeID']}):\n")
                for e in rowx['Errors']:
                    f.write(f"   - {e}\n")
                f.write("\n")
            f.write("Error Summary:\n")
            f.write(summary.to_string(index=False) + "\n\n")
            f.write(f"Total Rows Processed: {len(student_df)}\n")
            f.write(f"Total Rows with Errors: {len(error_df)}\n")
            f.write(f"Total Error Messages: {total_error_messages}  <-- graded count (must be < 50)\n")
        print(f"Errors saved to {log_filename}")

        print("\nSummary Report:")
        print(f"Total Rows Processed: {len(student_df)}")
        print(f"Total Rows with Errors: {len(error_df)}")
        print(f"Total Error Messages: {total_error_messages}  <-- graded count (must be < 50)")
    else:
        print("✅ No validation errors found! Great job!")
else:
    print("⚠ Could not perform validation because one or more DataFrames are invalid.")
````

Perfect data is **not always achievable**, and that’s okay! The goal is to **reduce errors as much as possible** while understanding where the data might still have gaps. Validation helps you:

- Identify areas that need improvement.
- Make informed decisions about **data quality**.
- Ensure your dataset is **as accurate and reliable as possible** for further analysis.

A dataset with **fewer than 50 errors** (the **Total Error Messages** line in your log) is generally acceptable, as long as the critical fields are accurate. Small inconsistencies are expected and should not discourage you from proceeding with analysis.

Rather than seeking absolute perfection, focus on **continuous improvement**—cleaner data leads to better insights! 🚀
