### 🔹 Database objects you will use
- **📂 Schema:** a folder of tables inside a database. Yours is the default schema, `public`.
- **📊 Table:** rows (records) and columns (fields).
- **🔑 Primary Key (PK):** the column that uniquely identifies each row. Every table has exactly one.
- **🔗 Foreign Key (FK):** a column that points to the primary key of another table. This is what draws the lines on an ERD.
- **🛑 Constraints:** rules the database enforces:
  - `NOT NULL`: the column can never be empty.
  - `UNIQUE`: no two rows can have the same value.
  - `CHECK`: values must pass a condition, e.g. `salary > 0`.
  - `DEFAULT`: the value used when none is given.

### 🔹 PostgreSQL data types

| Data Type | Description | Example |
| --- | --- | --- |
| `INTEGER` | Whole numbers | `100`, `-50` |
| `BIGINT` | Large whole numbers | `9000000000` |
| `SERIAL` | Auto-incrementing whole number, for ID primary keys | `1, 2, 3...` |
| `VARCHAR(n)` | Text up to `n` characters (pgAdmin calls it `character varying`) | `'Austin'` |
| `TEXT` | Text of any length | `'A long description...'` |
| `CHAR(n)` | Text of exactly `n` characters | `'TX'` |
| `BOOLEAN` | True/False | `TRUE`, `FALSE` |
| `DATE` | A date only | `'1971-12-10'` |
| `TIMESTAMP` | Date and time | `'2026-10-16 23:59:00'` |
| `NUMERIC(p,s)` | Exact decimal number: `p` digits in total, `s` after the point | `NUMERIC(10,2)` holds up to `99999999.99` |

**Choosing between similar types**
- **VARCHAR(n) vs TEXT:** `VARCHAR(n)` rejects anything longer than `n`, so pick `n` with room to spare. `TEXT` never rejects for length, good for descriptions.
- **NUMERIC vs INTEGER:** money has cents, so use `NUMERIC`. `INTEGER` is for whole numbers such as IDs.
- **DATE vs VARCHAR for dates:** always `DATE`. A date stored as text cannot be compared or used to compute ages.
- **IDs that look like numbers but are not math** (SSN `149-51-0794`, ZIP `02134`, phone `(512)555-6789`) are stored as `VARCHAR`. As numbers they would lose dashes and leading zeros.

## 📊 Step 3: Practice ERD, a small library database (not graded, not submitted)

This warm-up teaches every click you need in Step 4 on a tiny example. Do it in a **separate practice database** so it never mixes with `EmployeeDB`. The pgAdmin layout differs a little between versions; if a button is in a slightly different place, that is fine.

### 🔹 Open the ERD tool
1. Right-click **Databases** → **Create** → **Database...**, name it `librarydb`, **Save**.
2. Right-click **librarydb** → **ERD For Database**. A new ERD tab opens with an empty canvas.

### 🔹 Table 1: `books`
1. Click **Add Table** (top toolbar).
2. **General** tab, **Name:** `books`.
3. **Columns** tab, click **+** once per column:
   - `book_id`, type `serial`, switch **Primary key?** on
   - `title`, type `character varying`, switch **Not NULL?** on
   - `author`, type `character varying`, switch **Not NULL?** on
   - `available`, type `boolean`, **Default:** `true`
4. Click **OK** (or **Save**).

### 🔹 Table 2: `members`
1. **Add Table**, **Name:** `members`.
2. **Columns:**
   - `member_id`, `serial`, **Primary key?** on
   - `full_name`, `character varying`, **Not NULL?** on
   - `email`, `character varying`
   - `join_date`, `date`
3. **Make `email` unique:** **Constraints** tab → **Unique** → **+** → open the new row's edit (pencil) → **Definition** → **Columns:** pick `email`.
4. **OK**.

### 🔹 Table 3: `borrowed_books` (links books and members)
1. **Add Table**, **Name:** `borrowed_books`.
2. **Columns:**
   - `borrow_id`, `serial`, **Primary key?** on
   - `book_id`, `integer` (not `serial`: it stores a value copied from `books`, it does not generate its own)
   - `member_id`, `integer`
   - `borrow_date`, `date`
   - `return_date`, `date`
3. **Foreign keys:** **Constraints** tab → **Foreign Key** → **+** → edit (pencil) → **Columns** tab:
   - **Local column:** `book_id`, **References:** `books`, **Referencing:** `book_id` → **Add**
4. Add a second foreign key the same way: `member_id` → `members` → `member_id`.
5. **OK**. Two lines now connect `borrowed_books` to `books` and `members`.

> 💡 Many pgAdmin versions also have a **One-to-Many link** button on the toolbar that creates the same foreign key by picking the two tables. Either way is fine.

### 🔹 Tidy, generate, run
1. **Auto align** (toolbar) to arrange the tables. Double-click any table to edit it again.
2. **Generate SQL** (toolbar). pgAdmin opens a **Query Tool** tab containing `CREATE TABLE ...` statements.
3. Click **Execute** (▶ or F5). You should see **"Query returned successfully"**.
4. In the Object Explorer: **librarydb > Schemas > public > Tables**, right-click **Tables** → **Refresh**. Your 3 tables are there.

You now know every move needed for the real thing.

## 🏗️ Step 4: Design the EmployeeDB ERD (the graded part)

Your Assignment 3 file `Cleaned_data.csv` has 24 columns about each of roughly 14,300 employees. Storing it as one wide table would repeat each department description, division address and job description thousands of times. Instead you split it into **7 tables** in Third Normal Form (3NF): every fact is stored once, and tables point to each other with foreign keys.

Right-click **EmployeeDB** → **ERD For Database**, and build these tables.

### 🔹 The 7 tables and their columns (use these names exactly)

| Table | Columns, in this order | Where the values come from in `Cleaned_data.csv` |
|---|---|---|
| `personaldetails` | `ssn`, `name`, `honorific`, `highestdegree`, `email`, `phonenumber`, `birthdate`, `perstreet`, `percity`, `perstate`, `perzip` | One row per person. `honorific` is not in the CSV, it stays empty |
| `departments` | `departmentid`, `departmentname`, `departmentdescription` | The 6 departments |
| `jobtitles` | `jobtitleid`, `jobtitle` | The distinct job titles |
| `divisions` | `divisionid`, `divtype` | The 5 divisions (`South Division`, ...) |
| `divisionlocations` | `locationid`, `divisionid`, `divstreet`, `divcity`, `divstate`, `divzip`, `divisionphone` | One address per division |
| `jobdescriptions` | `jobdescid`, `jobtitleid`, `divisionid`, `description` | The description of each job title in each division |
| `employeedetails` | `emp_id`, `employeeid`, `ssn`, `departmentid`, `jobtitleid`, `divisionid`, `salary` | One row per employee, linking a person to their department, job and division. `emp_id` is a new auto-number; `employeeid` is KLB's own EmployeeID from the CSV |

> ⚠️ **Type every table and column name in lowercase, exactly as above.** If you type `PersonalDetails`, pgAdmin stores it as a case-sensitive quoted name `"PersonalDetails"`, and Assignment 5's loader, which writes `PersonalDetails` without quotes, then reports `relation "personaldetails" does not exist` and loads nothing. We tested this.

### 🔹 Required rules (Assignment 5 will not load your data otherwise)
1. **The first column of each table is its primary key.** The six `...id` keys (`departmentid`, `jobtitleid`, `divisionid`, `locationid`, `jobdescid`, `emp_id`) are `serial`. `ssn` is the key of `personaldetails` and is `character varying(11)`.
2. **Foreign-key columns use `integer`, not `serial`,** and `employeedetails.ssn` uses the same type as `personaldetails.ssn`.
3. **`birthdate` is `date`.** Text dates break Assignment 5's age query (−8 points).
4. **`salary` is `numeric(10,2)`** (or plain `numeric`). The largest salary is over $1,000,000, so `numeric(8,2)` overflows.
5. **Do NOT make `email`, `phonenumber`, `name` or `employeeid` UNIQUE.** In real life emails and employee IDs would be unique, but Assignment 3 allows up to 50 validator errors, so even a file that passes can contain a few repeats (our own reference file has 12 repeated emails), and different employees share phone numbers and names. One duplicate makes the whole Assignment 5 load stop.
6. **Leave `honorific` nullable** (no NOT NULL). The cleaned CSV has no honorific column, so it is always empty.
7. **SSN, ZIP and phone columns are text (`character varying`),** never numbers.
8. **`employeeid` is a plain `integer` column.** It keeps KLB's EmployeeID from your CSV (e.g. `436822`) so every row can be traced back to the company's records. It is **not** the primary key (`emp_id` is), so it gets no UNIQUE, no NOT NULL and no foreign key.

### 🔹 How long the text gets
For `character varying(n)`, choose `n` larger than the longest value below, with room to spare, or use `text`. A value longer than `n` makes the Assignment 5 load fail.

| Column(s) | Longest value in a correctly cleaned file |
|---|---|
| `ssn` | 11 (`149-51-0794`) |
| `name` | 33 |
| `highestdegree` | 10 (`Bachelors+`) |
| `email` | 34 |
| `phonenumber`, `divisionphone` | 13 (`(512)555-6789`) |
| `perstreet` | 36 |
| `percity` | 24 |
| `perstate`, `divstate` | 2 |
| `perzip`, `divzip` | 5 |
| `departmentname` | 15 (`Human Resources`) |
| `departmentdescription` | **104** (so `varchar(100)` fails, use `text`) |
| `jobtitle` | 29 |
| `divtype` | 16 (`Central Division`) |
| `divstreet` | 13 |
| `divcity` | 11 |
| `description` (job description) | **154** (use `text`) |

### 🔹 Your design choices (this is what the constraints criterion grades)
- **Foreign keys.** Decide which columns point to which table. Hint: a column with the same name as another table's primary key is a foreign key to that table. **When you are done there are exactly 7 relationship lines on your ERD.**
- **NOT NULL.** Mark the columns that must always have a value. At minimum: the names in the three lookup tables (`departmentname`, `jobtitle`, `divtype`), `personaldetails.name`, and in `employeedetails` the `ssn` and `salary`. Only add NOT NULL to a column if your `Cleaned_data.csv` has no blanks in it.
- **UNIQUE.** The lookup values `departmentname`, `jobtitle` and `divtype` should each appear only once. (Not `email`, `phonenumber`, `name` or `employeeid`, see rule 5.)
- **CHECK.** Add at least one. The natural one is `salary > 0`: in the table dialog, **Constraints** tab → **Check** → **+** → edit (pencil) → **Definition** → **Check:** `salary > 0`.

### 🔁 Need to change something and run it again?
The generated script uses `CREATE TABLE IF NOT EXISTS`, so running it a second time does **not** update tables that already exist. You only see notices like `relation "personaldetails" already exists, skipping`, your changes are silently ignored, and every foreign key gets added a second time. To start over cleanly, open a Query Tool on **EmployeeDB** and run:
```sql
DROP SCHEMA public CASCADE;
CREATE SCHEMA public;
```
This empties `EmployeeDB` only (there is no data in it yet). Then fix the ERD, **Generate SQL** again, save over `EmployeeDB_schema.sql`, and execute.

**Only ever run that reset inside `EmployeeDB`.** Check the Query Tool tab title first.
